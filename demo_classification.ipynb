```python
from os import listdir
from os.path import join, isdir
from syllabus_classificaiton.functions import extract
```

```python
base_dir = "input/Lehrveranstaltungen"
dest_dir = "output"

for main in listdir(base_dir):

    if isdir(join(base_dir, main)):
        print("\nWORKING IN AREA: ", main)
        for item in listdir(join(base_dir, main)):
            print("- Working on", item)
            with open(join(base_dir, main, item), "r", encoding="utf-8") as file:
                syllabus = file.read()
                syllabus = " ".join(syllabus.split())

            classification = extract(syllabus)

            text = "\n".join([item.name for item in classification.competences])

            with open(join(dest_dir, main, item), "w", encoding="utf-8") as file:
                file.write(text)
```
